## Drop the "Id" column

```python
df_new=df.drop("Id", axis=1)
df_new.head()
```

## For Numerical columns fill in missing data with median values

```python
df_new.dtypes
```

```python
df_new.isnull().sum()
```

```python
#LotFrontage
median_lf=df_new['LotFrontage'].median()
df_new['LotFrontage'].fillna(median_lf,inplace=True)

#MasVnrArea
median_lf=df_new['MasVnrArea'].median()
df_new['MasVnrArea'].fillna(median_lf,inplace=True)

#Converting GarageYrBlt to match YearBuilt. It makes sense to have them match since its the same type of reference.
df_new['GarageYrBlt']=df_new['GarageYrBlt'].astype('Int64')
median_lf=df_new['GarageYrBlt'].median()
df_new['GarageYrBlt'].fillna(median_lf,inplace=True)
```

# Categorical Columns: FIll in missing values with the mode
### Columns: Alley, MasVnrType, BsmtQual, BsmtCond, BsmtExposure, BsmtFinType1, BsmtFinType2, Electrical, FireplaceQu, GarageType, GarageFinish, GarageQual, GarageCond, PoolQC, Fence, MiscFeature

```python
mode_lf=df_new['Alley'].mode()
df_new['Alley'].fillna(mode_lf,inplace=True)

#MasVnrType
mode_lf=df_new['MasVnrType'].mode()
df_new['MasVnrType'].fillna(mode_lf,inplace=True)

#BsmtQual
mode_lf=df_new['BsmtQual'].mode()
df_new['BsmtQual'].fillna(mode_lf,inplace=True)

#BsmtCond
mode_lf=df_new['BsmtCond'].mode()
df_new['BsmtCond'].fillna(mode_lf,inplace=True)

#BsmtExposure
mode_lf=df_new['BsmtExposure'].mode()
df_new['BsmtExposure'].fillna(mode_lf,inplace=True)

#BsmtFinType1
mode_lf=df_new['BsmtFinType1'].mode()
df_new['BsmtFinType1'].fillna(mode_lf,inplace=True)

#BsmtFinType2
mode_lf=df_new['BsmtFinType2'].mode()
df_new['BsmtFinType2'].fillna(mode_lf,inplace=True)

#Electrical
mode_lf=df_new['Electrical'].mode()
df_new['Electrical'].fillna(mode_lf,inplace=True)

#FireplaceQu
mode_lf=df_new['FireplaceQu'].mode()
df_new['FireplaceQu'].fillna(mode_lf,inplace=True)

#GarageType
mode_lf=df_new['GarageType'].mode()
df_new['GarageType'].fillna(mode_lf,inplace=True)

#GarageFinish
mode_lf=df_new['GarageFinish'].mode()
df_new['GarageFinish'].fillna(mode_lf,inplace=True)

#GarageQual
mode_lf=df_new['GarageQual'].mode()
df_new['GarageQual'].fillna(mode_lf,inplace=True)

#GarageCond
mode_lf=df_new['GarageCond'].mode()
df_new['GarageCond'].fillna(mode_lf,inplace=True)

#PoolQC
mode_lf=df_new['PoolQC'].mode()
df_new['PoolQC'].fillna(mode_lf,inplace=True)

#Fence
mode_lf=df_new['Fence'].mode()
df_new['Fence'].fillna(mode_lf,inplace=True)

#MiscFeature
mode_lf=df_new['MiscFeature'].mode()
df_new['MiscFeature'].fillna(mode_lf,inplace=True)
```

## Convert Categorical Variables to Dummy Variables

```python
#Applying just a general get_dummies to the entire frame. This applies it to columns where the dtype is object, string, or category.
df_dummies=pd.get_dummies(df_new)
df_dummies.head()
```

## Split Data into a Train and Test Set

```python
from sklearn.model_selection import train_test_split
x=df_dummies.drop(columns=['SalePrice'])
y=df_dummies['SalePrice']
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

## Run a Linear Regression and Report the R2 and RMSE of the Test Set

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, y_train)
```

```python
from sklearn.metrics import r2_score

y_pred_test=model.predict(x_test)
r2_test=r2_score(y_test,y_pred_test)
print(r2_test)
```

```python
from sklearn.metrics import root_mean_squared_error as RMSE

y_pred_test=model.predict(x_test)
rmse_test=RMSE(y_test,y_pred_test)
print(rmse_test)
```

## Fit and Transform the Training Features with a PCA so that 90% of the variance is retained

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

scaler=StandardScaler()
x_train_scale=scaler.fit_transform(x_train)

pca=PCA(n_components=0.9, whiten=True)
x_train_pca=pca.fit_transform(x_train_scale)

print("Original number of features:", x_train.shape[1])
print("Reduced number of features:",x_train_pca.shape[1])
```

## Transform but don't fit the test data to the same PCA 

```python
x_test_scale=scaler.transform(x_test)
x_test_pca=pca.transform(x_test_scale)

print("Original number of features:", x_test.shape[1])
print("Reduced number of features:",x_test_pca.shape[1])
```

## Run a LInear Regression with the PCA data

```python
model = LinearRegression()
model.fit(x_train_pca, y_train)
```

```python
y_pred_test=model.predict(x_test_pca)
r2_test=r2_score(y_test,y_pred_test)
print(r2_test)
```

```python
y_pred_test=model.predict(x_test_pca)
rmse_test=RMSE(y_test,y_pred_test)
print(rmse_test)
```

## Take your original original training features and apply a min-max scaler to them

```python
from sklearn.preprocessing import MinMaxScaler

scaler=MinMaxScaler()
x_train_minmax=scaler.fit_transform(x_train)

#The following operation raises this error. I can print the scaled training data but the original data exceeds the limit I think.
#print("Original Training Data:", x_train)
#print("Scaled Training Data:",x_train_minmax)
```

## Find the MinMax scaled features in the training data with a variance above 0.1

```python
from sklearn.feature_selection import VarianceThreshold
thresholder=VarianceThreshold(threshold=0.1)
features_variance=thresholder.fit_transform(x_train_minmax)

features_variance[0:3]
```

## Transform but don't fit the Test data with the MinMax Scaler and Variance Threshold

```python
x_test_minmax=scaler.transform(x_test)

thresholder=VarianceThreshold(threshold=0.1)
features_variance=thresholder.fit_transform(x_test_minmax)

features_variance[0:3]
```

## Apply the linear regression model, r2, and rmse to the high variance threshold.

```python
model = LinearRegression()
model.fit(x_train_minmax, y_train)
```

```python
y_pred_test=model.predict(x_test_minmax)
r2_test=r2_score(y_test,y_pred_test)
print(r2_test)
```

```python
y_pred_test=model.predict(x_test_minmax)
rmse_test=RMSE(y_test,y_pred_test)
print(rmse_test)
```

## Summaerize Your Findings
By applying the MinMax Scaler and the Variance Threshold of 0.1 the model it improved. The R squared value for the minmax/threshold option rose by about 3% and it lowered the RMSE by roughly $4000. By using the threshold we told the model to only consider things that accounted for a lot of the variance and ignore things similar to our target feature.

# Categorical Feature Selection
## Import thr data and ensure it is loaded correctly

```python
df=pd.read_csv("mushrooms.csv")
df.head()
```

## Convert all of the categorical variables to dummy variables

```python
#This seelct everything except the class column since that is our target variable.
target_column='class'
df_dummies=pd.get_dummies(df.drop(columns=[target_column]),drop_first=True)
df_final=pd.concat([df_dummies,df[[target_column]]],axis=1)
column_to_move=df_final.pop('class')
df_final.insert(0,'class',column_to_move)
df_final.head()
```

## Split the Data into Training and Test Set

```python
#I chose to make the class_e(edible) the target variable.
x=df_final.drop(columns=['class'])
y=df_final['class']
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
y=encoder.fit_transform(y)
print(y)
```

```python
x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=42)
```

## Fit a decision tree classifier to the training data

```python
from sklearn.tree import DecisionTreeClassifier
model=DecisionTreeClassifier()
model.fit(x_train,y_train)
```

## Report the accuracy and create a confusion matrix on the test set

```python
pred_y=model.predict(x_test)
from sklearn.metrics import accuracy_score
accuracy=accuracy_score(y_test,pred_y)
print(accuracy)
```

```python
from sklearn.metrics import confusion_matrix
matrix=confusion_matrix(y_test,pred_y)
print(matrix)
```
